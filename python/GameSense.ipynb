## OUTLIERS

```python
numeric_cols = [
    'age',
    'playtime_hours',
    'sleep_hours',
    'energy_drinks',
    'screen_brightness',
    'apm',
    'heart_rate_bpm',
    'hrv_score',
    'reaction_time_ms'
]

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print(f"\n{col}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
```

```python
## make sleep groups
```

```python
df['sleep_groups'] = pd.cut(
    df['sleep_hours'],
    bins=[0, 5, 7, 9, 12],
    labels=['<5 Hours', '5-7 Hours', '7-9 Hours', '9+ Hours']
)
```

```python
sleep_win_rate = pd.crosstab(
    df['sleep_groups'],
    df['match_outcome'],
    normalize='index'
) * 100

sleep_win_rate.round(2)
```

```python
df['sleep_groups'].value_counts().sort_index()
```

```python
# Create APM groups
df['apm_groups'] = pd.cut(
    df['apm'],
    bins=[0, 100, 200, 300, 400, 600],
    labels=['<100', '100-200', '200-300', '300-400', '400+']
)
```

```python
# Calculate Win/Loss percentage by APM group
apm_win_rate = pd.crosstab(
    df['apm_groups'],
    df['match_outcome'],
    normalize='index'
) * 100
```

```python
# Display Win/Loss percentages
print("Win/Loss percentage by APM group:")
print(apm_win_rate.round(2))
```

```python
# Display number of players in each APM group
print("\nNumber of players in each APM group:")
print(df['apm_groups'].value_counts().sort_index())
```

```python
# Create Reaction Time groups
df['reaction_time_group'] = pd.cut(
    df['reaction_time_ms'],
    bins=[0, 220, 250, 280, 310, 350],
    labels=['<220 ms', '220-250 ms', '250-280 ms', '280-310 ms', '310+ ms']
)
```

```python
# Calculate Win/Loss percentage
reaction_win_rate = pd.crosstab(
    df['reaction_time_group'],
    df['match_outcome'],
    normalize='index'
) * 100
```

```python
# Display Win/Loss percentages
print("Win/Loss percentage by Reaction Time group:")
print(reaction_win_rate.round(2))
```

```python
# Display number of players
print("\nNumber of players in each Reaction Time group:")
print(df['reaction_time_group'].value_counts().sort_index())
```

```python
# Calculate Win/Loss percentage by Energy Drinks
energy_win_rate = pd.crosstab(
    df['energy_drinks'],
    df['match_outcome'],
    normalize='index'
) * 100
```

```python
# Display Win/Loss percentages
print("Win/Loss percentage by Energy Drinks:")
print(energy_win_rate.round(2))
```

```python
# Display number of players in each group
print("\nNumber of players in each Energy Drinks group:")
print(df['energy_drinks'].value_counts().sort_index())
```
